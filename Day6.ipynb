```python
import pandas as pd

df = pd.read_csv('students_cleaned.csv')

print('Given Dataset= ')
print(df)

#Printing specific columns
print(df['Name'])
print(df['Maths'])
print(df[['Name','Maths','Science']])

#Filtering Rows
print('\nStudents Scoring above 80 in Maths= ')
print(df[df['Maths']>80][['Name','Maths']])

print('\nStudents Scoring above 90 in Science= ')
print(df[df['Science']>90][['Name','Science']])

print('\nStudents who are female= ')
print(df[df['Gender'] == 'Female'])

print('\nStudents who are older than 20= ')
print(df[df['Age']>20])

#Multiple conditions for filtering rows
print('\nStudents Scored above 80 in Maths and 85 in Science= ')
print(df[(df['Maths']>80) & (df['Science']>85)])

print('\nStudents Scored above 85 in English and are female= ')
print(df[(df['English']>85) & (df['Gender'] == 'Female')])

#Sorting
print('\nStudents sorted by Maths marks (Ascending)= ')
print(df.sort_values(by='Maths'))

print('\nStudents sorted by Maths marks (Descending)= ')
print(df.sort_values(by='Maths', ascending=False))

#Selecting Top Students
print('\nTop 5 marks scored in Maths by Students= ')
print(df.sort_values(by='Maths', ascending = False).head())

#Mini Project
print('\nTotal Female Students = ', len(df[df['Gender'] == 'Female']))

print('\nTotal Male Students = ', len(df[df['Gender'] == 'Male']))

print('\nStudents scoring below 70 in English= ')
print(df[df['English'] < 70][['Name', 'English']])

print('\nTop 3 Students in Maths= ')
print(df.sort_values(by='Maths', ascending=False).head(3))

print('\nBottom 3 Students in Science= ')
print(df.sort_values(by='Science').head(3))
```
