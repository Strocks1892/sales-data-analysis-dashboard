```python
import pandas as pd

df = pd.read_csv('students_cleaned.csv')

#Basic Dataset Summary
print('Dataset Shape= ',df.shape)
print('\nColumn Names= ')
print(df.columns)

print('\nStatistical Summary= ')
print(df.describe())

#Average Marks
df['Average']=(df['Maths']+df['Science']+df['English'])/3
print(df[['Name','Average']])

#Top Performer
top_student = df.loc[df['Average'].idxmax()]

print('Top Performer= ')
print(top_student)

#Lowest Performer
lowest_student = df.loc[df['Average'].idxmin()]

print('Lowest Performer= ')
print(lowest_student)

#Subject wise avg
print('Average Maths= ', df['Maths'].mean())
print('Average Science= ', df['Science'].mean())
print('Average English= ', df['English'].mean())

#Count students by gender
print(df['Gender'].value_counts())

#Students above avg
overall_avg = df['Average'].mean()

print(df[df['Average'] > overall_avg][['Name', 'Average']])

#Mini Project
print('Total Students =', len(df))

print('Highest Maths Marks =', df['Maths'].max())

print('Highest Science Marks =', df['Science'].max())

print('Highest English Marks =', df['English'].max())

print('Average Marks of Class =', df['Average'].mean())

print('Students Above Class Average ')
print(df[df['Average'] > df['Average'].mean()][['Name', 'Average']])

print('Top 3 Performers= ')
print(df.sort_values(by='Average', ascending=False)[['Name', 'Average']].head(3))
```
