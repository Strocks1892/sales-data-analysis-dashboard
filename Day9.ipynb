```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("students_cleaned.csv")

#Scatter plot
sns.scatterplot(data=df, x="Maths", y="Science")

plt.title("Maths vs Science Marks")
plt.show()

#Count plot
sns.countplot(data=df, x="Gender")

plt.title("Gender Count")
plt.show()

#Box plot
sns.boxplot(data=df, y="Maths")

plt.title("Maths Marks Distribution")
plt.show()

#Histogram
sns.histplot(data=df, x="English", bins=5)

plt.title("English Marks Distribution")
plt.show()

#Correlation Heatmap
corr = df[["Maths", "Science", "English"]].corr()

sns.heatmap(corr, annot=True, cmap="Blues")

plt.title("Subject Correlation")
plt.show()

#Mini Project
# Average Marks
df["Average"] = (
    df["Maths"] +
    df["Science"] +
    df["English"]
) / 3

plt.figure(figsize=(8,5))

sns.barplot(data=df, x="Name", y="Average")

plt.title("Average Marks of Students")
plt.xticks(rotation=45)

plt.show()
```
