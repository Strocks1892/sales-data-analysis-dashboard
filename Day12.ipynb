```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("sales_data.csv")

#Dataset Overview
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nSummary:")
print(df.describe())

#Buisness Insights

print("Total Revenue =", df["Sales"].sum())

print("Total Orders =", len(df))

print("Average Revenue =", df["Sales"].mean())

print("Best Selling Product =")
print(df.groupby("Product")["Sales"].sum().sort_values(ascending=False))

print("Best Performing Region =")
print(df.groupby("Region")["Sales"].sum().sort_values(ascending=False))

print("Highest Revenue Order =")
print(df.loc[df["Sales"].idxmax()])

#Dashboard Charts
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="Product", y="Sales")

plt.xticks(rotation=45)

plt.title("Product-wise Sales")

plt.show()

plt.figure(figsize=(6,5))

sns.barplot(
    x=df.groupby("Region")["Sales"].sum().index,
    y=df.groupby("Region")["Sales"].sum().values
)

plt.title("Region-wise Sales")

plt.show()

plt.figure(figsize=(7,5))

sns.histplot(df["Sales"], bins=6)

plt.title("Sales Distribution")

plt.show()

plt.figure(figsize=(7,5))

sns.boxplot(y=df["Quantity"])

plt.title("Quantity Distribution")

plt.show()

plt.figure(figsize=(6,5))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```
