```python
# Variables and Datatypes
name = "Stavan"
age = 20
height = 5.9
is_student = True

print(name)
print(age)
print(height)
print(is_student)

# Operators
a = 15
b = 4

print(a + b)
print(a % b)
print(a > b)

# Conditions
marks = 82

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
else:
    print("Grade C")

# Loops
for i in range(1, 6):
    print(i)

count = 1
while count <= 5:
    print(count)
    count += 1

# Functions
def greet(name):
    return f"Hello, {name}!"

print(greet("Stavan"))

# Mini challenge
no_of_students = int(input('Enter the no. of students= '))
for i in range(no_of_students):
    name_of_student = input('Enter the name of Student= ')
    maths = int(input('Enter marks of student score in maths= '))
    sci = int(input('Enter marks of student score in  science= '))
    eng = int(input('Enter marks of student score in english= '))
    if maths<0 or maths>100 or sci<0 or sci>100 or eng<0 or eng>100:
        print("Enter valid marks range(0-100)")
    else:
        total = sci+eng+maths
        avg = round(total/3,2)
        print(f"Student Name : {name_of_student}")
        print(f"Total Marks  : {total}")
        print(f"Average      : {avg}")
        if avg < 35:
            print(f"{name_of_student} has failed.")
        elif avg < 50:
            print(f"{name_of_student} passed with Grade D")
        elif avg < 65:
            print(f"{name_of_student} passed with Grade C")
        elif avg < 75:
            print(f"{name_of_student} passed with Grade B")
        elif avg < 85:
            print(f"{name_of_student} passed with Grade A")
        elif avg <= 100:
            print(f"{name_of_student} passed with Grade A+")
```
