```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('students_cleaned.csv')

#Basic Informations
print(df.head())

print(df.info())

print(df.describe())

df["Average"] = (
    df["Maths"] +
    df["Science"] +
    df["English"]
) / 3

print("Total Students =", len(df))

print("Class Average =", df["Average"].mean())

print("Highest Average= ")
print(df.sort_values(by="Average", ascending=False)[["Name","Average"]].head(1))

print("Lowest Average= ")
print(df.sort_values(by="Average")[["Name","Average"]].head(1))

print("Students above 80 average= ")
print(df[df["Average"] > 80][["Name","Average"]])

print("Students above 80 average and are female= ")
print(df[
    (df["Gender"]=="Female") &
    (df["Average"]>80)
][["Name","Average"]])
print("Subject wise highest marks= ")
print("Maths =", df["Maths"].max())
print("Science =", df["Science"].max())
print("English =", df["English"].max())

#Visualization
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="Name", y="Average")

plt.xticks(rotation=45)

plt.title("Average Marks")

plt.show()

#Correlation
sns.heatmap(
    df[["Maths","Science","English","Average"]].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.show()

#Histogram
sns.histplot(df["Average"], bins=5)

plt.title("Average Marks Distribution")

plt.show()

df["Grade"] = ""

df.loc[df["Average"] >= 90, "Grade"] = "A+"
df.loc[(df["Average"] >= 80) & (df["Average"] < 90), "Grade"] = "A"
df.loc[(df["Average"] >= 70) & (df["Average"] < 80), "Grade"] = "B"
df.loc[(df["Average"] >= 60) & (df["Average"] < 70), "Grade"] = "C"
df.loc[(df["Average"] >= 50) & (df["Average"] < 60), "Grade"] = "D"
df.loc[df["Average"] < 50, "Grade"] = "F"

print(df[["Name", "Average", "Grade"]])
```
