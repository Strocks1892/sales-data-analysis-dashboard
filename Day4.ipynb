```python
import pandas as pd 

#load CSV File
df = pd.read_csv('students.csv')

print(df)

print('First 5 Rows:')
print(df.head())

print('First 3 Rows:')
print(df.head(3))

print('Last 5 Rows:')
print(df.tail())

print('Last 3 Rows:')
print(df.tail(3))

print('Information Regarding Dataset= ')
print(df.info())

print('No. of Rows and Columns',df.shape)
print('Name of Columns= ',df.columns)
print('Describe the Dataset= ',df.describe())

#Mini Project
print('Total Students = ', len(df))

print('Average Maths Marks = ', df['Maths'].mean())

print('Highest Science Marks = ', df['Science'].max())

print('Lowest English Marks = ', df['English'].min())

print('Students whose marks are greater than 80 in Maths= ',df[df['Maths']>80]['Name'])

print('Students whose marks are greater than 90 in Science= ',df[df['Science']>90]['Name'])
```
