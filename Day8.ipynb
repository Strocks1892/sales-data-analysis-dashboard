```python
import pandas as pd 
import matplotlib.pyplot as plt

df = pd.read_csv('students_cleaned.csv')

#Bar Chart
plt.figure(figsize=(10,5))
plt.bar(df["Name"], df["Maths"])

plt.title("Maths Marks of Students")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()

#Line Chart
plt.figure(figsize=(10,5))
plt.plot(df["Name"], df["Science"], marker="o")

plt.title("Science Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()

#Pie Chart
gender = df["Gender"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    gender,
    labels=gender.index,
    autopct="%1.1f%%"
)

plt.title("Gender Distribution")

plt.show()

#Histogram
plt.figure(figsize=(8,5))
plt.hist(df["Maths"], bins=5)

plt.title("Distribution of Maths Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()

#Scatter Plot
plt.figure(figsize=(8,5))
plt.scatter(df["Maths"], df["Science"])

plt.title("Maths vs Science")
plt.xlabel("Maths")
plt.ylabel("Science")

plt.show()

#Mini Project
# Average Marks
df["Average"] = (
    df["Maths"] +
    df["Science"] +
    df["English"]
) / 3

plt.figure(figsize=(10,5))
plt.bar(df["Name"], df["Average"])
plt.title("Average Marks")
plt.xlabel("Students")
plt.ylabel("Average")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10,5))
plt.plot(df["Name"], df["Average"], marker="o")
plt.title("Average Marks")
plt.xlabel("Students")
plt.ylabel("Average")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
